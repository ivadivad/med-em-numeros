## Convertendo `AnoMes` para data

`AnoMes` chega como inteiro no formato `AAAAMM` (ex: `202201` = janeiro de
2022). Pra usar como eixo de tempo depois, converto pra `datetime` com
`pd.to_datetime(..., format="%Y%m")` e já ordeno por essa coluna.

```python
df["Data"] = pd.to_datetime(df["AnoMes"], format="%Y%m")
df = df.sort_values("Data").reset_index(drop=True)
df[["AnoMes", "Data"]].head()
```

## Lacunas e duplicatas

Duas checagens antes de confiar na série: (1) nenhum mês repetido, (2)
nenhum mês faltando entre o primeiro e o último.

```python
duplicados = df["AnoMes"].duplicated().sum()
print(f"meses duplicados: {duplicados}")
```

```python
esperado = pd.date_range(df["Data"].min(), df["Data"].max(), freq="MS")
faltando = sorted(set(esperado) - set(df["Data"]))

print(f"intervalo: {df['Data'].min():%Y-%m} a {df['Data'].max():%Y-%m}")
print(f"meses esperados: {len(esperado)}  |  meses presentes: {len(df)}")
print(f"meses faltando: {len(faltando)}")
faltando
```

**Resultado: 0 meses duplicados, 0 meses faltando.** A série vai de janeiro
de 2022 a abril de 2026, 52 meses seguidos, uma linha por mês — exatamente o
que o `fraude_cobertura.csv` do coletor já indicava, agora confirmado a
partir do CSV consolidado mesmo, sem depender do relatório da coleta.
